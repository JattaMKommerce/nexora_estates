```python
%%writefile /content/voices.py
# 5. Her voice for every sentence (Kokoro, af_bella), with calm silence at both ends
import json, os, numpy as np, soundfile as sf
from kokoro import KPipeline
clips = json.load(open('/content/clips.json'))
os.makedirs('/content/wav', exist_ok=True)
pipe = KPipeline(lang_code='a')
pad = np.zeros(int(24000 * 0.35), dtype=np.float32)
for i, c in enumerate(clips):
    out = '/content/wav/%s.wav' % c['id']
    if os.path.exists(out): continue
    audio = np.concatenate([np.asarray(a) for _, _, a in pipe(c['say'], voice='af_bella', speed=0.95)])
    sf.write(out, np.concatenate([pad, audio, pad]), 24000)
    print('%d/%d voice %s' % (i + 1, len(clips), c['id']), flush=True)
sf.write('/content/wav/idle.wav', np.zeros(24000 * 12, dtype=np.float32), 24000)   # 12 s idle loop
print('VOICES OK')
```

```python
%%writefile /content/videos.py
# 6. A talking video for every sentence (Ditto), each easing in and out of her resting pose
import os, sys, json, copy, subprocess, traceback
os.chdir('/content/ditto'); sys.path.insert(0, '/content/ditto')
from stream_pipeline_offline import StreamSDK
from inference import run
CFG, ROOT = './checkpoints/ditto_cfg/v0.4_hubert_cfg_pytorch.pkl', './checkpoints/ditto_pytorch'
MORE = {'setup_kwargs': {'fade_type': 's', 'fade_out_keys': ('exp', 'pitch', 'yaw', 'roll', 't')},
        'run_kwargs': {'fade_in': 8, 'fade_out': 8}}
clips = json.load(open('/content/clips.json'))
jobs = [('idle', 'idle')] + [(c['id'], c['id']) for c in clips]
os.makedirs('/content/raw', exist_ok=True); os.makedirs('/content/web', exist_ok=True)
sdk = StreamSDK(CFG, ROOT)
failed = []
for i, (cid, wav) in enumerate(jobs):
    raw, web = '/content/raw/%s.mp4' % cid, '/content/web/%s.mp4' % cid
    if os.path.exists(web): continue
    try:
        try:
            run(sdk, '/content/wav/%s.wav' % wav, '/content/hannah.jpg', raw, more_kwargs=copy.deepcopy(MORE))
        except Exception:
            traceback.print_exc(); sdk = StreamSDK(CFG, ROOT)       # start Ditto afresh and try once more
            run(sdk, '/content/wav/%s.wav' % wav, '/content/hannah.jpg', raw, more_kwargs=copy.deepcopy(MORE))
        # web version: 1280 wide H.264, starts playing before it has fully downloaded
        audio = ['-an'] if cid == 'idle' else ['-c:a', 'aac', '-b:a', '96k']
        subprocess.run(['ffmpeg', '-loglevel', 'error', '-y', '-i', raw, '-vf', 'scale=1280:-2', '-c:v', 'libx264',
                        '-preset', 'slow', '-crf', '23', '-pix_fmt', 'yuv420p'] + audio + ['-movflags', '+faststart', web], check=True)
        print('%d/%d %s ok' % (i + 1, len(jobs), cid), flush=True)
    except Exception:
        traceback.print_exc(); failed.append(cid)
print('FAILED:', failed if failed else 'none')
print('VIDEOS OK' if not failed else 'SOME VIDEOS FAILED: send a screenshot')
```
